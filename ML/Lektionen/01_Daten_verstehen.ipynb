## A — 14.10.: Tabelle und Python
Ziel: Beobachtung, Merkmal und Datentyp unterscheiden. Eine Zeile repräsentiert eine Beobachtung. Spalten beschreiben Eigenschaften. Ein DataFrame organisiert die Tabelle; eine ausgewählte Spalte ist eine Series.
Führe jede Zelle aus. Schreibe vor der Ausgabe auf, was du erwartest.

```python
import pandas as pd
from io import StringIO
import matplotlib.pyplot as plt

csv_text = """haushalt,kwh
A,4
B,6
C,
D,8
D,8
E,-2
F,40
"""
raw = pd.read_csv(StringIO(csv_text))
raw
```

```python
print(raw.shape)
print(raw.dtypes)
print(raw.head())
verbrauch = raw['kwh']
print(verbrauch)
```

### Python-Brücke
Eine Liste enthält Elemente. Eine for-Schleife durchläuft diese Elemente. Die Bedingung entscheidet, welche Werte berücksichtigt werden. Die folgende Miniübung nutzt absichtlich eine separate Liste ohne fehlende Werte.

```python
werte = [4, 6, -2, 8]
plausibel = []
for wert in werte:
    if wert >= 0:
        plausibel.append(wert)
print(plausibel)
```

**Deine Erklärung:** Was wird ausgeschlossen und aufgrund welcher Annahme? Ändere einen Wert und sage das Ergebnis vorher.

## B — 16.10.: Qualität prüfen
Fehlend heißt unbekannt. 0 ist dagegen ein gemessener Wert. Duplikate werden anhand der fachlichen Bedeutung erkannt. Hier soll ein Haushalt genau einmal vorkommen. Ein auffälliger positiver Wert kann echt sein; Auffälligkeit allein rechtfertigt kein Löschen.

```python
print('Fehlende Werte je Spalte:')
print(raw.isna().sum())
print('Wiederholte Zeilen nach der ersten:', raw.duplicated().sum())
print('Negative Werte:')
print(raw.loc[raw['kwh'] < 0])
```

**Selbstprüfung:** Warum sind Haushalte als Zahlencode kein sinnvoller Messwert? Warum reicht `dtypes` nicht aus, um fachliche Plausibilität zu prüfen?

## C — 19.10.: Nachvollziehbar bereinigen
Bewahre raw unverändert auf. Eine Arbeitskopie macht Entscheidungen nachvollziehbar. In diesem Übungsbeispiel entfernen wir die exakt wiederholte Zeile und behandeln negative Messwerte als unbekannt. In realen Daten wären Quelle und Messprozess zu klären.

```python
clean = raw.drop_duplicates().copy()
clean.loc[clean['kwh'] < 0, 'kwh'] = float('nan')
clean
```

**Wichtig:** Fehlwerte bleiben hier fehlend. Weder automatisches Auffüllen mit 0 noch pauschales Entfernen aller betroffenen Haushalte wird als allgemeine Regel gelernt.

## D — 21.10.: Statistik und Grafik
Der Mittelwert ist Summe geteilt durch Anzahl. Der Median beschreibt die Mitte der sortierten Werte. Ein hoher Wert wirkt stärker auf den Mittelwert. Beide Kennzahlen beschreiben unterschiedliche Aspekte; keine ist immer besser. pandas lässt fehlende Werte bei diesen Berechnungen standardmäßig aus; berichte deshalb die verwendete Anzahl.

```python
print('Verwendete Messwerte:', clean['kwh'].count())
print('Mittelwert:', clean['kwh'].mean())
print('Median:', clean['kwh'].median())
measured = clean.dropna(subset=['kwh'])
ax = measured.plot.bar(x='haushalt', y='kwh', legend=False, color='#147D83')
ax.set_ylabel('Tagesverbrauch (kWh)')
ax.set_title('Nur vorhandene, nichtnegative Messwerte')
plt.tight_layout()
plt.show()
```

**Deine Interpretation:** Welche Haushalte fehlen in der Grafik und warum? Was kann die Grafik nicht über deren Verbrauch aussagen?
